```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
print("Shape:", df.shape)

df.info()
```

```python
df.isnull().sum()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```

```python
df.isnull().sum()
```

```python
df = df.drop("customerID", axis=1)

df.head()
```

```python
df = pd.get_dummies(df, drop_first=True)

df.head()
```

```python
df.shape
```

```python
# Features
X = df.drop("Churn_Yes", axis=1)

# Target
y = df["Churn_Yes"]

print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train, y_train)

log_pred = log_model.predict(X_test)

print("Logistic Regression Accuracy:")
print(accuracy_score(y_test, log_pred))

print("\nClassification Report:")
print(classification_report(y_test, log_pred))
```

```python
tree_model = DecisionTreeClassifier(random_state=42)

tree_model.fit(X_train, y_train)

tree_pred = tree_model.predict(X_test)

print("Decision Tree Accuracy:")
print(accuracy_score(y_test, tree_pred))

print("\nClassification Report:")
print(classification_report(y_test, tree_pred))
```

```python
print("Logistic Regression Accuracy:",
      accuracy_score(y_test, log_pred))

print("Decision Tree Accuracy:",
      accuracy_score(y_test, tree_pred))
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": tree_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance.head(3)
```

## Business Summary

The analysis shows that customers with month-to-month contracts are more likely to leave the company than those with long-term contracts. Customers who have been with the company for a shorter period also tend to churn more frequently. Monthly charges have an impact on customer churn, with higher-paying customers showing a higher chance of leaving. These insights can help the company improve customer retention by offering better long-term plans and personalized offers to high-risk customers.
