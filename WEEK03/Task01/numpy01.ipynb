```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("train.csv")

df.head()
```

```python
print("Shape of Dataset:", df.shape)

df.info()
```

```python
# Fill missing Age values with median
df["Age"].fillna(df["Age"].median(), inplace=True)

# Fill missing Embarked values with mode
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)

# Drop Cabin column because it has too many missing values
df.drop("Cabin", axis=1, inplace=True)
```

```python
df.drop(["PassengerId", "Name", "Ticket"], axis=1, inplace=True)

df.head()
```

```python
df = pd.get_dummies(df, columns=["Sex", "Embarked"], drop_first=True)

df.head()
```

```python
# Features (Input)
X = df.drop("Survived", axis=1)

# Target (Output)
y = df["Survived"]

print(X.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
print(classification_report(y_test, y_pred))
```

## Model Evaluation

The Logistic Regression model achieved an accuracy of **80.97%**.

The classification report shows that the model performs slightly better at predicting passengers who did not survive than those who survived.

The model has a Precision of **0.79**, Recall of **0.73**, and F1-Score of **0.76** for the survived class, indicating that some survivors were missed by the model.

```python
param_grid = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["liblinear", "lbfgs"]
}
```

```python
grid = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring="accuracy"
)

grid.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best Cross Validation Score:", grid.best_score_)
```

```python
best_model = grid.best_estimator_

best_model
```

```python
y_pred_tuned = best_model.predict(X_test)
```

```python
tuned_accuracy = accuracy_score(y_test, y_pred_tuned)

print("Tuned Accuracy:", tuned_accuracy)
```

```python
print(confusion_matrix(y_test, y_pred_tuned))
```

```python
print(classification_report(y_test, y_pred_tuned))
```

```python
comparison = pd.DataFrame({
    "Model": ["Original", "Tuned"],
    "Accuracy": [
        accuracy_score(y_test, y_pred),
        accuracy_score(y_test, y_pred_tuned)
    ]
})

comparison
```

GridSearchCV tested different combinations of hyperparameters and selected the best configuration.

In this case, the tuned model performed similarly to the original model. This shows that the original model was already well configured, and hyperparameter tuning confirmed that only minor adjustments were needed.

Accuracy alone can be misleading because it does not show how well a model predicts each class. In an imbalanced dataset, a model may achieve high accuracy by predicting only the majority class while failing to identify the minority class. Metrics like Precision, Recall, and F1-Score provide a more complete evaluation of model performance.

Hyperparameter tuning helped identify the best combination of model parameters automatically. In this project, the tuned model achieved performance similar to the original model, indicating that the initial model was already close to optimal.

## Model Evaluation After Hyperparameter Tuning

After applying GridSearchCV, the Logistic Regression model achieved an accuracy of **81.72%**, which is slightly better than the original model.

The tuned model improved Precision, meaning it made fewer incorrect positive predictions. Although Recall decreased slightly, the overall model performance became more balanced.

This demonstrates how hyperparameter tuning can optimize model performance by selecting the best parameter combination automatically.
